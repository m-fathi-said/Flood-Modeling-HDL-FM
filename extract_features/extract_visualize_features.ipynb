#### Import libraries

```python
# Global admin libraries
import gc
from pathlib import Path

# Data manipulation and visualization libraries
import numpy as np
import matplotlib.pyplot as plt
import torch
```

#### Global settings

```python
# Make paths cross-platform compatible with pathlib
data_dir = Path(DATA_DIR)
fstore_dir = Path(FSTORE_DIR)

# Ensure the paths exist before using
if data_dir.exists():
    print(f'Data directory found: {data_dir}')
else:
    print(f'WARNING: Data directory not found at {data_dir}!')

# Create directory to save data if it doesn't exist
if fstore_dir.exists():
    print(f"Directory for saving final dataset found: {fstore_dir}")
else:
    fstore_dir.mkdir(parents=True, exist_ok=True)  # Ensure directory exists, create if not
    print(f"WARNING: Directory for saving preprocessed features not found at {fstore_dir}. Created directory.")
```

#### X_train

```python
# Load the tensor
X_train_tensor = torch.load(data_dir / 'train_x.pt')
print(f'X_train_tensor occupies: {X_train_tensor.nbytes / (1024 ** 3):.2f} GB')
X_train_tensor.shape, X_train_tensor.dtype, X_train_tensor.device
```

```python
# Select a sample index to visualize (e.g., the first sample)
sample_idx = 100

fig, axes = plt.subplots(1, 4, figsize=(16, 4))

for i in range(4):
    ax = axes[i]
    im = ax.imshow(X_train_tensor[sample_idx, i, :, :].cpu().numpy(), cmap='terrain')
    ax.set_title(f'Channel {i}')
    ax.axis('off')
    plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)

plt.tight_layout()
plt.show()
```

```python
# Save features as compressed numpy array for sharing and as uncompressed for partial memory loading
np.savez_compressed(fstore_dir / "X_Qb_train.npz", X_Qb_train=X_train_tensor[:, 0, :, :].numpy())
np.save(fstore_dir / "X_Qb_train.npy", X_train_tensor[:, 0, :, :].numpy())

np.savez_compressed(fstore_dir / "X_elev_train.npz", X_elev_train=X_train_tensor[:, 1, :, :].numpy())
np.save(fstore_dir / "X_elev_train.npy", X_train_tensor[:, 1, :, :].numpy())

np.savez_compressed(fstore_dir / "X_h_lag_train.npz", X_h_lag_train=X_train_tensor[:, 2, :, :].numpy())
np.save(fstore_dir / "X_h_lag_train.npy", X_train_tensor[:, 2, :, :].numpy())

np.savez_compressed(fstore_dir / "X_h_train.npz", X_h_train=X_train_tensor[:, 3, :, :].numpy())
np.save(fstore_dir / "X_h_train.npy", X_train_tensor[:, 3, :, :].numpy())

del X_train_tensor  # Delete the tensor to free up memory
gc.collect()  # Force immediate garbage collection

print(f'Saved data to {fstore_dir} and cleared from memory.')
```

#### Memory-efficient loading example: Load data as memory mapped numpy arrays for partial loading

This is an example of how to load select parts of big data without it occupying masses of memory. Set the load_into_mem variable to True and look at the spike in your memory usage on Task Manger (Windows) or Activity Monitor (Mac). Set it to False and look again. You should see that the spike is larger when the array is loaded into memory.

You can also check what impact the number of samples that you are loading has by experimenting with different values for num_samples (has to be lower than the max number in X_elev_train_temp though).

```python
# Decide how many samples to load (number must be smaller than len(X_train_temp))
num_samples = 900

# Decide whether to load into memory or not
load_into_mem = False  # Set to True to load into memory, False to keep memory mapping

if load_into_mem:
    # Load the features as memory-mapped arrays and copy them into memory
    X_Qb_train_temp = np.load(fstore_dir / "X_Qb_train.npy", mmap_mode='r')[:num_samples].copy()  # Feature 1
    X_elev_train_temp = np.load(fstore_dir / "X_elev_train.npy", mmap_mode='r')[:num_samples].copy()  # Feature 2
    X_h_lag_train_temp = np.load(fstore_dir / "X_h_lag_train.npy", mmap_mode='r')[:num_samples].copy()  # Feature 3
    X_h_train_temp = np.load(fstore_dir / "X_h_train.npy", mmap_mode='r')[:num_samples].copy()  # Feature 4
else:
    # Load the features as memory-mapped arrays
    X_Qb_train_temp = np.load(fstore_dir / "X_Qb_train.npy", mmap_mode='r')[:num_samples]  # Feature 1
    X_elev_train_temp = np.load(fstore_dir / "X_elev_train.npy", mmap_mode='r')[:num_samples]  # Feature 2
    X_h_lag_train_temp = np.load(fstore_dir / "X_h_lag_train.npy", mmap_mode='r')[:num_samples]  # Feature 3
    X_h_train_temp = np.load(fstore_dir / "X_h_train.npy", mmap_mode='r')[:num_samples]  # Feature 4

del X_Qb_train_temp, X_elev_train_temp, X_h_lag_train_temp, X_h_train_temp  # Delete from memory again
gc.collect()  # Force immediate garbage collection
```

If you perform operations on the memory mapped arrays such as concatenation, these will be automatically loaded into memory at the time of operation. Run the following cell with load_into_mem=True, check the spike in your memory, then run for load_into_mem=False and check again. 

You will find that when the data is not loaded into memory directly, the operation is faster. You might see that the peak memory usage is actually a bit larger when the data is only loaded into memory at the time of operation due to some computational overhead. If there is no computational overhead, the peak memory usage should be the same for both cases.

To avoid this memory overhead at the time of operation on memory mapped arrays, load the data as memory mapped arrays first to keep memory low, slice them to the desired number of samples, then copy them into memory for further operations. Otherwise, keep them as memory mapped arrays until they are used in an operation for speed and efficient memory allocation.

In addition, when dealing with big data, it is advisable to delete variables with 'del' and gc.collect() throughout the code when they are no longer needed. For .py files, all data is removed from memory when the file finishes running, so this is only necessary for intermediate steps but not at the end. For jupyter notebooks, variables are kept in memory until the kernel is reset, so it is advisable to delete variables at the end of a cell that are no longer needed when the cell in question has finished running.

```python
# Decide how many samples to load (number must be smaller than len(X_train_temp))
num_samples = 900

# Decide whether to load into memory or not
load_into_mem = False  # Set to True to load into memory, False to keep memory mapping

if load_into_mem:
    # Load the features as memory-mapped arrays and copy them into memory
    X_Qb_train_temp = np.load(fstore_dir / "X_Qb_train.npy", mmap_mode='r')[:num_samples].copy()  # Feature 1
    X_elev_train_temp = np.load(fstore_dir / "X_elev_train.npy", mmap_mode='r')[:num_samples].copy()  # Feature 2
    X_h_lag_train_temp = np.load(fstore_dir / "X_h_lag_train.npy", mmap_mode='r')[:num_samples].copy()  # Feature 3
    X_h_train_temp = np.load(fstore_dir / "X_h_train.npy", mmap_mode='r')[:num_samples].copy()  # Feature 4
else:
    # Load the features as memory-mapped arrays
    X_Qb_train_temp = np.load(fstore_dir / "X_Qb_train.npy", mmap_mode='r')[:num_samples]  # Feature 1
    X_elev_train_temp = np.load(fstore_dir / "X_elev_train.npy", mmap_mode='r')[:num_samples]  # Feature 2
    X_h_lag_train_temp = np.load(fstore_dir / "X_h_lag_train.npy", mmap_mode='r')[:num_samples]  # Feature 3
    X_h_train_temp = np.load(fstore_dir / "X_h_train.npy", mmap_mode='r')[:num_samples]  # Feature 4

# Concatenate the features along the channel dimension (operations will automatically load data into memory)
X_train = np.concatenate((X_Qb_train_temp[:, np.newaxis, :, :],
                          X_elev_train_temp[:, np.newaxis, :, :],
                          X_h_lag_train_temp[:, np.newaxis, :, :],
                          X_h_train_temp[:, np.newaxis, :, :]), axis=1)

# Check data
print(f"X_train has shape: {X_train.shape} and occupies: {X_train.nbytes / (1024 ** 2):.0f} MB (on disk)")
X_train.shape, X_train.dtype, X_train.min(), X_train.mean(), X_train.max()

# Select a sample index to visualize (e.g., the first sample)
sample_idx = 0

fig, axes = plt.subplots(1, 4, figsize=(16, 4))

for i in range(4):
    ax = axes[i]
    im = ax.imshow(X_train[sample_idx, i, :, :], cmap='terrain')
    ax.set_title(f'Channel {i}')
    ax.axis('off')
    plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)

plt.tight_layout()
plt.show()

del X_train, X_Qb_train_temp, X_elev_train_temp, X_h_lag_train_temp, X_h_train_temp  # Delete from memory again
gc.collect()  # Force immediate garbage collection
```
